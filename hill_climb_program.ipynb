```python

def hill_climbing():
    def f(x):
        return -(x - 5) ** 2 + 25

    x = 0
    while True:
        if f(x + 1) > f(x):
            x = x + 1
        elif f(x - 1) > f(x):
            x = x - 1
        else:
            break

    print("Best Position:", x)
    print("Maximum Value:", f(x))

hill_climbing()
```
